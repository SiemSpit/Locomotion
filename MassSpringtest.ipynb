```python
import numpy as np
import matplotlib.pyplot as plt
import taichi as ti
import os

ti.init(arch=ti.cpu)
```

```python
m = 1.0
k = 1.0
c = 0.1

u0 = 0.1
ut0 = 0.0
utt0 = 0.0

per = 2 * np.pi * np.sqrt(m / k)
dt = per / 100
total_step = 1000
```

```python
# time integration

ulist = []
utlist = []
uttlist = []
for i in range(total_step):
    if i == 0:
        u = u0
        ut = ut0
        utt = utt0
    else:
        utt = -k * u / m - c * ut / m
        ut += utt * dt
        u += ut * dt
    ulist.append(u)
    utlist.append(ut)
    uttlist.append(utt)
```

```python
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.plot(ulist)
plt.title('u')
plt.subplot(1, 3, 2)
plt.plot(utlist)
plt.title('ut')
plt.subplot(1, 3, 3)
plt.plot(uttlist)
plt.title('utt')
plt.tight_layout()
plt.show()
```
